```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
np.random.seed(42)
student_df=pd.read_csv(r"D:\learingAndDev\Own_application\Gen_AI_Learnings\Learinings\ML_BASICS\Data_science_library\data_sets\house_prices_realistic_1000.csv")
student_df.columns.to_list()
```

```python
from sklearn.model_selection import train_test_split

## select feature 
X_feature=student_df[["bedrooms","bathrooms","sqft","floors","school_rating"]]
Y_output=student_df[["price"]]

## split data and train model 
X_train,X_test,Y_train,Y_test=train_test_split(X_feature,Y_output,test_size=0.2,random_state=42)
print(X_train[0:1])
```

```python
model=LinearRegression()
model.fit(X_train,Y_train)

print(model.coef_)
print("-q")
print(model.intercept_)

pred=model.predict(X_test)
print(pred)
```
